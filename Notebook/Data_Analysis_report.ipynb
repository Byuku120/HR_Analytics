```python
# Prepare data again
attrition_by_role = HR_Analytics.groupby(['JobRole', 'Attrition']).size().reset_index(name='count')
total_per_role = attrition_by_role.groupby('JobRole')['count'].transform('sum')
attrition_by_role['percentage'] = round((attrition_by_role['count'] / total_per_role) * 100, 2)

# Focus only on 'Yes' attrition to compare percentage by role
attrition_yes = attrition_by_role[attrition_by_role['Attrition'] == 'Yes'].copy()
```

- Test for Normality (Shapiro-Wilk)
I use the Shapiro-Wilk test to check normality of attrition percentages

```python
from scipy.stats import shapiro

stat, p_value = shapiro(attrition_yes['percentage'])

print("Shapiro-Wilk Test for Normality")
print(f"Statistic: {stat:.4f}, p-value: {p_value:.4f}")
```

The Shapiro-Wilk test shows a p-value of 0.3134, which is above 0.05, indicating the data is normally distributed. This means it’s appropriate to use parametric tests for analysis.


- Chi-Square Test of Independence: Attrition by Job Role

To determine whether attrition status is related to job role, we performed a Chi-Square Test of Independence.

```python
# Create a contingency table of JobRole vs Attrition
contingency_table = pd.crosstab(HR_Analytics['JobRole'], HR_Analytics['Attrition'])

print("Contingency Table:")
print(contingency_table)

# Perform Chi-Square Test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Test of Independence")
print(f"Chi2 Statistic: {chi2:.4f}")
print(f"Degrees of Freedom: {dof}")
print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print(" Significant relationship between Job Role and Attrition (reject null hypothesis).")
else:
    print(" No significant relationship between Job Role and Attrition (fail to reject null hypothesis).")
```

# Contingency Table
This table summarizes how many employees in each job role either left ("Yes") or stayed ("No").

so, p-value < 0.05  We reject the null hypothesis, job role and attrition are significantly related.

- The correct statistical approach (Chi-Square Test) was used to determine the relationship between job role and attrition. The results should now guide decisions related to workforce retention strategies, focusing on specific roles if the relationship is found to be significant.

- Check if there is any correlation

```python
# Correlation between percentage and job role
attrition_yes['JobRole_Code'] = attrition_yes['JobRole'].astype('category').cat.codes
corr_coef, corr_p = stats.spearmanr(attrition_yes['percentage'], attrition_yes['JobRole_Code'])

print("\nCorrelation Analysis (Spearman)")
print(f"Correlation coefficient: {corr_coef:.4f}, p-value: {corr_p:.4f}")

if corr_p < 0.05:
    print("Significant correlation exists between job role and attrition percentage.")
else:
    print("No significant correlation between job role and attrition percentage.")
```


# Correlation Analysis

- Spearman's ρ (rho) is used as it captures monotonic relationships.
- A significant p-value (< 0.05) indicates a meaningful correlation.

```python
from scipy.stats import shapiro

# Step 1: Test Normality of attrition percentages across departments using Shapiro-Wilk test
print("Normality Test (Shapiro-Wilk) for Attrition Percentage by Department:")

stat, p_value = shapiro(attrition_yes['percentage'])
print(f"Test Statistic = {stat:.4f}, p-value = {p_value:.4f}")

if p_value > 0.05:
    print("Data appears to be normally distributed. Use parametric tests.\n")
else:
    print("Data is NOT normally distributed. Use non-parametric tests.\n")
```

This result means that the percentages of attrition by department follow a normal distribution. The p-value is 0.3134, which is greater than 0.05, so we do not reject the assumption of normality. Because the data looks normally distributed, it’s appropriate to use parametric tests for further analysis.


```python
# Let's create contingency table for Dept vs Attrition (Yes/No) for correlation test
contingency_table = pd.crosstab(HR_Analytics['Department'], HR_Analytics['Attrition'])

print("Contingency Table (Department vs Attrition):")
print(contingency_table)

# Chi-Square test for independence between Department and Attrition
chi2, p_chi, dof, expected = chi2_contingency(contingency_table)

print(f"\nChi-Square Test for Independence between Department and Attrition:")
print(f"Chi2 Statistic = {chi2:.4f}, p-value = {p_chi:.4f}")

if p_chi < 0.05:
    print("There is a significant association between Department and Attrition.\n")
else:
    print("No significant association between Department and Attrition.\n")
```

This table displays the number of employees who remained (No) and those who left (Yes) within each department. The Chi-square test indicates a significant relationship between department and attrition, as the p-value (0.0045) is below 0.05. This suggests that employee attrition rates vary across different departments.

```python
# Step 4: Correlation between Department and Attrition?
# Since Department is categorical, and Attrition is categorical, correlation can be measured by Cramér's V
def cramers_v(confusion_matrix):
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    r, k = confusion_matrix.shape
    return np.sqrt(chi2 / (n * (min(k-1, r-1))))

cramers_v_value = cramers_v(contingency_table)
print(f"Cramér's V (measure of association strength) = {cramers_v_value:.4f}")
```

- Recommendation 

The analysis shows that department does affect attrition, as indicated by a statistically significant association (p = 0.0045), but the effect is weak (Cramér’s V = 0.0856). This means that while department plays a role, it is not the main driver of employee turnover. Therefore, this finding should serve as a starting point for deeper analysis—such as examining job roles within departments, evaluating employee satisfaction, work-life balance, or performance levels. It’s also advisable to focus retention efforts on departments with higher attrition rates, like Sales or Research & Development.

```python
# Step 1: Create contingency table of Gender vs Attrition
contingency_table = pd.crosstab(HR_Analytics['Gender'], HR_Analytics['Attrition'])

print("Contingency Table:\n", contingency_table)

# Step 2: Perform Chi-Square test of independence
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"\nChi-Square Test Result:")
print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"P-value: {p_value:.4f}")

# Step 3: Significance interpretation
alpha = 0.05
if p_value < alpha:
    print("Result: Reject the null hypothesis — Gender and Attrition are significantly associated.")
else:
    print("Result: Fail to reject the null hypothesis — No significant association between Gender and Attrition.")

# Step 4: Cramér's V — Strength of association
def cramers_v(chi2, n, shape):
    return np.sqrt(chi2 / (n * (min(shape)-1)))

n = contingency_table.to_numpy().sum()
cramers_v_value = cramers_v(chi2, n, contingency_table.shape)

print(f"\nCramér's V: {cramers_v_value:.4f}")

# Step 5: Interpret Cramér's V
if cramers_v_value < 0.10:
    strength = "negligible"
elif cramers_v_value < 0.30:
    strength = "weak"
elif cramers_v_value < 0.50:
    strength = "moderate"
else:
    strength = "strong"

print(f"Interpretation: The strength of the association between Gender and Attrition is {strength}.")
```

This analysis shows that there is no significant relationship between gender and attrition. The Chi-square test gave a p-value of 0.2928, which is higher than 0.05, meaning we cannot say gender affects whether employees leave or stay. Additionally, Cramér's V value is very low (0.0274), indicating the association between gender and attrition is negligible. In simple terms, gender does not meaningfully influence employee turnover in this data.


- I would like to recommend that Gender doesn’t significantly affect attrition, so efforts to reduce turnover should focus else where like Job role and age group, but keep monitoring gender equity in different areas like in how they are both paying if they are equally and if they promote them equally.

```python
from scipy.stats import shapiro
# Step 1: Prepare data — extract performance rating for both OverTime groups
yes_overtime = HR_Analytics[HR_Analytics['OverTime'] == 'Yes']['PerformanceRating']
no_overtime = HR_Analytics[HR_Analytics['OverTime'] == 'No']['PerformanceRating']

# Step 2: Test for normality using Shapiro-Wilk test (on both groups)
print("Normality Test (Shapiro-Wilk):")
stat_yes, p_yes = shapiro(yes_overtime)
stat_no, p_no = shapiro(no_overtime)

print(f"OverTime = Yes: stat = {stat_yes:.4f}, p = {p_yes:.4f}")
print(f"OverTime = No : stat = {stat_no:.4f}, p = {p_no:.4f}")
```

Both p-values are less than 0.05, so performance ratings are not normally distributed in either group.

- Let's use mannwhitneyu() non-parametric test for comparing two independent groups

```python
from scipy.stats import shapiro 
stat, p = mannwhitneyu(yes_overtime, no_overtime, alternative='two-sided')
print(f"Mann-Whitney U test:\nStatistic = {stat:.4f}, p-value = {p:.4f}")

if p < 0.05:
    print(" Statistically significant difference in performance rating between OverTime groups.")
else:
    print(" No significant difference in performance rating between OverTime groups.")
```

This result means that there is no meaningful difference in performance ratings between employees who work overtime and those who do not.

The p-value is 0.8506, which is much higher than the typical threshold of 0.05. This tells us that any small differences in performance are likely due to random chance, not because of overtime status.

```python
# Step 4: Correlation between OverTime and PerformanceRating
# OverTime is binary categorical → convert to numeric: Yes=1, No=0
HR_Analytics['OverTime_Num'] = HR_Analytics['OverTime'].map({'Yes': 1, 'No': 0})

# Use Point-Biserial Correlation (special case of Pearson for binary vs continuous)
corr, p_corr = pointbiserialr(HR_Analytics['OverTime_Num'], HR_Analytics['PerformanceRating'])

print(f"\nPoint-Biserial Correlation between OverTime and PerformanceRating:")
print(f"Correlation = {corr:.4f}, p-value = {p_corr:.4f}")

if p_corr < 0.05:
    print(" There is a statistically significant correlation between OverTime and PerformanceRating.")
else:
    print(" No significant correlation between OverTime and PerformanceRating.")
```

- The point-biserial correlation result shows a correlation value of 0.0049 with a p-value of 0.8506, which means there is no significant relationship between working overtime and performance rating. The correlation is very close to zero, indicating that whether an employee works overtime or not has no meaningful impact on their performance rating in this dataset.
